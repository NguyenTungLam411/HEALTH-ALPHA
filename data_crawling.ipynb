```python
import pandas as pd
from ucimlrepo import fetch_ucirepo

# Download the Pima Indians Diabetes dataset
diabetes = fetch_ucirepo(id=17)

# Features and target
X = diabetes.data.features
y = diabetes.data.targets

# Combine into one DataFrame
df = pd.concat([X, y], axis=1)

# Display sample records
print(df.head())

# Dataset information
print("\nDataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

# Save data to CSV
df.to_csv("diabetes_sample.csv", index=False)

print("\nSaved to diabetes_sample.csv")
```

```python
import pandas as pd

url = (
    "https://raw.githubusercontent.com/"
    "jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"
)

columns = [
    "Pregnancies", "Glucose", "BloodPressure",
    "SkinThickness", "Insulin", "BMI",
    "DiabetesPedigreeFunction", "Age", "Outcome"
]

df = pd.read_csv(url, names=columns)

print(df)
df.to_csv("diabetes.csv", index=False)
```
